```python
import pandas as pd
import time
from dataretrieval import nwis, waterdata

# Define the San Joaquin Valley HUC8 subbasins
sjv_huc8_list = ['18040001', '18040002', '18040003','18040006', '18040007','18040008','18040009','18040011','18040014','18040051','18040010','18040012','18040013']
start_date = '2000-01-01'
end_date = '2020-12-31'
param_code = '00060'

all_station_data = []

# Loop through each subbasin
for huc in sjv_huc8_list:
    print(f"\nScanning HUC8: {huc}...")

    # Get all active gauges
    sites_df, sites_md = nwis.what_sites(huc=huc, parameterCd=param_code, hasDataTypeCd='dv')

    if sites_df.empty:
        continue

    site_ids = sites_df['site_no'].tolist()

    # get Drainage Area
    site_info_df, info_md = nwis.get_info(sites=site_ids)

    # Sort by drainage area and pick the top one
    valid_sites = site_info_df.dropna(subset=['drain_area_va'])

    if valid_sites.empty:
        continue

    outlet_gauge = valid_sites.sort_values(by='drain_area_va', ascending=False).iloc[0]
    site_no = outlet_gauge['site_no']
    station_name = outlet_gauge['station_nm']

    print(f"   Found Outlet Gauge: USGS-{site_no}")
    print(f"   Name: {station_name}")
    print(f"   Drainage Area: {outlet_gauge['drain_area_va']} sq miles")

    # get daily data and location
    lat = outlet_gauge['dec_lat_va']
    lon = outlet_gauge['dec_long_va']

    try:
        df, md = waterdata.get_daily(
            monitoring_location_id=f"USGS-{site_no}",
            parameter_code=param_code,
            time=f"{start_date}/{end_date}"
        )
        if not df.empty:
            # inject the coordinates into the dataset
            df['latitude'] = lat
            df['longitude'] = lon
            all_station_data.append(df)

    except Exception as e:
        print(f"   Failed to download data: {e}")
```

```python
# unit conversion factor
cfs_to_m3s = 0.0283168 # cfs to m^3/s
#build dictionary
gauge_dict = {df['monitoring_location_id'].iloc[0]: df for df in all_station_data}

# Middle San Joaquin-Lower Chowchilla (18040001)
print("\n--- Middle San Joaquin-Lower Chowchilla ---")
if 'USGS-11273400' in gauge_dict:
    Middle_San_Joaquin_df = gauge_dict['USGS-11273400'].copy()
    Middle_San_Joaquin_df['discharge_m3s'] = Middle_San_Joaquin_df['value'] * cfs_to_m3s
    lat_val = Middle_San_Joaquin_df['latitude'].iloc[0]
    lon_val = Middle_San_Joaquin_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Middle_San_Joaquin_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary (Likely no data for 2000-2020).")

# Lower San Joaquin River (18040002)
print("\n--- Lower San Joaquin River ---")
if 'USGS-11274550' in gauge_dict:
    Lower_San_Joaquin_df = gauge_dict['USGS-11274550'].copy()
    Lower_San_Joaquin_df['discharge_m3s'] = Lower_San_Joaquin_df['value'] * cfs_to_m3s
    lat_val = Lower_San_Joaquin_df['latitude'].iloc[0]
    lon_val = Lower_San_Joaquin_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Lower_San_Joaquin_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# San Joaquin Delta (18040003)
print("\n--- San Joaquin Delta ---")
if 'USGS-11303500' in gauge_dict:
    San_Joaquin_Delta_df = gauge_dict['USGS-11303500'].copy()
    San_Joaquin_Delta_df['discharge_m3s'] = San_Joaquin_Delta_df['value'] * cfs_to_m3s
    lat_val = San_Joaquin_Delta_df['latitude'].iloc[0]
    lon_val = San_Joaquin_Delta_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(San_Joaquin_Delta_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper San Joaquin (18040006)
print("\n--- Upper San Joaquin ---")
if 'USGS-11247000' in gauge_dict:
    Upper_San_Joaquin_df = gauge_dict['USGS-11247000'].copy()
    Upper_San_Joaquin_df['discharge_m3s'] = Upper_San_Joaquin_df['value'] * cfs_to_m3s
    lat_val = Upper_San_Joaquin_df['latitude'].iloc[0]
    lon_val = Upper_San_Joaquin_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_San_Joaquin_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Fresno River (18040007)
print("\n--- Fresno River ---")
if 'USGS-11258000' in gauge_dict:
    Fresno_River_df = gauge_dict['USGS-11258000'].copy()
    Fresno_River_df['discharge_m3s'] = Fresno_River_df['value'] * cfs_to_m3s
    lat_val = Fresno_River_df['latitude'].iloc[0]
    lon_val = Fresno_River_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Fresno_River_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper Merced (18040008)
print("\n--- Upper Merced ---")
if 'USGS-11272500' in gauge_dict:
    Upper_Merced_df = gauge_dict['USGS-11272500'].copy()
    Upper_Merced_df['discharge_m3s'] = Upper_Merced_df['value'] * cfs_to_m3s
    lat_val = Upper_Merced_df['latitude'].iloc[0]
    lon_val = Upper_Merced_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_Merced_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper Tuolumne (18040009)
print("\n--- Upper Tuolumne ---")
if 'USGS-11290000' in gauge_dict:
    Upper_Tuolumne_df = gauge_dict['USGS-11290000'].copy()
    Upper_Tuolumne_df['discharge_m3s'] = Upper_Tuolumne_df['value'] * cfs_to_m3s
    lat_val = Upper_Tuolumne_df['latitude'].iloc[0]
    lon_val = Upper_Tuolumne_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_Tuolumne_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper Stanislaus (18040010)
print("\n--- Upper Stanislaus ---")
if 'USGS-11303000' in gauge_dict:
    Upper_Stanislaus_df = gauge_dict['USGS-11303000'].copy()
    Upper_Stanislaus_df['discharge_m3s'] = Upper_Stanislaus_df['value'] * cfs_to_m3s
    lat_val = Upper_Stanislaus_df['latitude'].iloc[0]
    lon_val = Upper_Stanislaus_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_Stanislaus_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper Calaveras California (18040011)
print("\n--- Upper Calaveras ---")
if 'USGS-11309500' in gauge_dict:
    Upper_Calaveras_df = gauge_dict['USGS-11309500'].copy()
    Upper_Calaveras_df['discharge_m3s'] = Upper_Calaveras_df['value'] * cfs_to_m3s
    lat_val = Upper_Calaveras_df['latitude'].iloc[0]
    lon_val = Upper_Calaveras_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_Calaveras_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper Mokelumne (18040012)
print("\n--- Upper Mokelumne ---")
if 'USGS-11325500' in gauge_dict:
    Upper_Mokelumne_df = gauge_dict['USGS-11325500'].copy()
    Upper_Mokelumne_df['discharge_m3s'] = Upper_Mokelumne_df['value'] * cfs_to_m3s
    lat_val = Upper_Mokelumne_df['latitude'].iloc[0]
    lon_val = Upper_Mokelumne_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_Mokelumne_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Upper Cosumnes (18040013)
print("\n--- Upper Cosumnes ---")
if 'USGS-11336000' in gauge_dict:
    Upper_Cosumnes_df = gauge_dict['USGS-11336000'].copy()
    Upper_Cosumnes_df['discharge_m3s'] = Upper_Cosumnes_df['value'] * cfs_to_m3s
    lat_val = Upper_Cosumnes_df['latitude'].iloc[0]
    lon_val = Upper_Cosumnes_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Upper_Cosumnes_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")

# Panoche-San Luis Reservoir (18040014)
print("\n--- Panoche-San Luis Reservoir ---")
if 'USGS-11255575' in gauge_dict:
    Panoche_San_Luis_df = gauge_dict['USGS-11255575'].copy()
    Panoche_San_Luis_df['discharge_m3s'] = Panoche_San_Luis_df['value'] * cfs_to_m3s
    lat_val = Panoche_San_Luis_df['latitude'].iloc[0]
    lon_val = Panoche_San_Luis_df['longitude'].iloc[0]
    print(f"  Location: {lat_val}, {lon_val}")
    print(Panoche_San_Luis_df[['time','discharge_m3s']].head())
else:
    print("Data missing from dictionary.")
```
